```python
def getIndex(num_list, target):
    # target이 리스트에 저장되어 있는 위치(index) 반환
    for i in range(len(num_list)):
        if num_list[i] == target:
            return i
    return -1


def getMax(num_list):
    # 리스트에서 가장 큰 숫자 반환
    max_num = num_list[0]

    for num in num_list:
        if num > max_num:
            max_num = num

    return max_num


def getMin(num_list):
    # 리스트에서 가장 작은 숫자 반환
    min_num = num_list[0]

    for num in num_list:
        if num < min_num:
            min_num = num

    return min_num


def countGT(num_list, target):
    # target보다 큰 수의 개수 반환
    count = 0

    for num in num_list:
        if num > target:
            count += 1

    return count


def sumList(num_list):
    # 리스트의 모든 값의 합 반환
    total = 0

    for num in num_list:
        total += num

    return total


def swapList(num_list):
    # 리스트를 역순으로 저장
    num_list.reverse()


number_list = [23, 45, 27, 11, 25, 65, 78]

print(getIndex(number_list, 25))
print(getMax(number_list))
print(getMin(number_list))
print(countGT(number_list, 42))
print(sumList(number_list))

swapList(number_list)
print(number_list)
```
